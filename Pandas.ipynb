```python
import pandas as pd
```

```python
data = {
    'Name': ['Vaibhav', 'Aryan', 'Rudra', 'Riyansh', 'XYZ'],
    'Age': [1, 2, 99, 50, 22],
    'City': ['New York', 'Los Angeles', 'Chicago', 'Houston', 'Phoenix']
}

df = pd.DataFrame(data)
print(df)
```

```python
print("first two rows of the dataframe are: ")
print(df.head(2))
print('\n')
print("last two rows of the dataframe are: ")
print(df.tail(2))
```

```python
print("eleminating a column from the dataframe: ")
print(df.drop("City", axis=1))
print("\n Eleminating a row from the dataframe: ")
print(df.drop(index=4, axis=0))
```

```python
print(df.loc[0]) 
```

```python
marks = pd.DataFrame({
    'Marks' : [-100, 100, 0, 50, 90]
})

df = pd.merge(df, marks, left_index=True, right_index=True)
print(df)
```

```python
print("Description of the DataFrame:", df.describe())
```

```python
print("Check for null values: ", df.isnull().sum())
```

```python
df['Grade'] = df['Marks'].apply(lambda x: 'A' if x < 0 else ('B' if x >= 100 else ('C' if x == 0 else ('D' if x == 50 else 'F'))))
print(df)
```

```python
df1 = pd.DataFrame({'a': [None, 3, 3], 'b' : [1, 2, None]})
# Fill null values with given 0 and inplace = True means, change in the original dataframe.
df1.fillna(0, inplace=True)
print(df1)
```

```python
df2 = pd.DataFrame({
    'Name': ['Vaibhav', 'Aryan', 'Rudra', 'Riyansh', 'Riyansh']
})

print("Drop duplicates from the DataFrame:")
print(df2.drop_duplicates())
```

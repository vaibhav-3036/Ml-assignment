```python
import numpy as np
```

```python
arr = np.array([1, 2, 3, 4, 5]) # 1D array
arr2 = np.array([[1, 2, 3], [4, 5, 6]]) # 2D array
arr3 = np.array([[[1, 2], [3, 4]], [[5, 6], [7, 8]]]) # 3D array
```

```python
# Elements in the array
print(arr.size)
print(arr2.size)
print(arr3.size)

print("\n")
# Dimensions of the array
print(arr.ndim)
print(arr2.ndim)
print(arr3.ndim)
```

```python
print("Shape", arr.shape)

print("Reshaping the array")
arr.reshape(5, 1)
```

```python
arr_copy = arr.copy()
print("Original array:", arr)
print("Copied array:", arr_copy)
```

```python
demo = arr2.ravel()
print("Ravelled array:", demo)
demo[0] = 10
print("Modified ravelled array:", demo)
```

```python
print(arr.sum()) # Sum of all elements in the array
print(arr2.sum(axis=1))  # Sum of each row in the array
```

```python
#Broadcasting
print(arr*3) # Each element will be multiplied by 3
```

```python
# Masking

print(arr[arr > 2]) # Elements greater > 2
```

```python
# Where

print(np.where(arr > 2)) # positions of elements greater than 2
```

```python
marks = np.array([67, 45, 23, 76, 98, 45, 98, 44, 90, 95, 34, 67, 65, 34, 89, 79, 87, 80, 76])

grades = np.select(
    [marks >= 90, marks >= 70, marks >= 50, marks >= 33],
    ["A", "B", "C", "D"],
    default="F"
)

print("Grades for the marks are:")
for mark, grade in zip(marks, grades):
    print(f"Mark: {mark} - Grade: {grade}")
```
